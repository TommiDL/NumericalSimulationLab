```python
import pandas as pd
import matplotlib.pyplot as plt
import os
```

```python
os.listdir()
```

```python
for el in os.listdir():
    if ('.csv'  not in el) or ('M' not in el):
        continue
    print(el)
    M=int(el.split('M')[1].split('.')[0])
        
    df=pd.read_csv(el)
    plt.title(f'M={M}')
    plt.errorbar(
        df.throws,df.avg-0.5,yerr=df.err, 
        #c='r', ecolor='b'
    )

    plt.hlines(0, 0, df.throws[len(df)-1]+1, ls='--', colors='r')
    plt.xlabel('#throws')
    plt.ylabel('<r>-1/2')
    plt.grid(True)
    plt.show()
```

```python
for el in os.listdir():
    if ('.csv'  not in el) or ('M' not in el):
        continue
    print(el)
    M=int(el.split('M')[1].split('.')[0])
        
    df=pd.read_csv(el)

    plt.hist(df.err, label=f'M={M}')
plt.legend()
plt.show()
```

```python

df=pd.read_csv('data/var_estimation_M100000.csv')
plt.errorbar(
    df.throws,df.avg-1/12,yerr=df.err, 
    #c='r', ecolor='b'
)

plt.hlines(0, 0, df.throws[len(df)-1]+1, ls='--', colors='r')
plt.xlabel('#throws')
plt.ylabel('<(r-0.5)^2>-1/12')
plt.grid(True)
plt.show()
```

# Chi 2

```python
import pandas as pd 
import matplotlib.pyplot as plt
import numpy as np
```

```python
df=pd.read_csv('data/chi2.csv')

df
```

```python
np.sqrt(df.data.var())
```

```python
plt.plot(df.data, '*')
plt.hlines(
    100, 0, len(df),
    ls='--', colors='r', 
    label='# of intervals'
)

plt.hlines(
    df.data.mean(), 0, len(df), 
    ls='--', colors='g', 
    label=f'avg={"%.1f"%df.data.mean()}'
)

plt.fill_between(
    range(0,101), 
    df.data.mean(), 
    df.data.mean()+np.sqrt(df.data.var()),
    color='r', alpha=0.5
)
plt.fill_between(
    range(0,101), 
    df.data.mean(), 
    df.data.mean()-np.sqrt(df.data.var()),
    color='r', alpha=0.5
)


plt.ylabel(r'$\chi^2$')
plt.xlabel('distribution index')
plt.grid(True)
plt.legend()
plt.show()
```
