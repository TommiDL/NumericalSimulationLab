## Integral with uniform distribution

```python
integral=pd.read_csv('data/cosinteg_M1000000.csv')

plt.title('Uniform distribution')

plt.errorbar(
    integral.throws, integral.avg, 
    color='r', ecolor='c', 
    yerr=integral.err
)
plt.hlines(1, 0, 1000000, linestyles='--', colors='r')

plt.ylabel(r'<$\int_0^1 \frac{\pi}{2} cos(\pi x/2 )dx$>')
plt.xlabel('# throws')
plt.grid(True)
plt.show()
```

```python
df_is=pd.read_csv('data/cosinteg_M1000000_is.csv')
plt.title('Importance sampling (first order taylor expansion)')


plt.errorbar(df_is.throws, df_is.avg, color='r', ecolor='c', yerr=df_is.err)
plt.hlines(1, 0, 1000000, linestyles='--', colors='r')

plt.ylabel(r'<$\int_0^1 \frac{\pi}{2} cos(\pi x/2 )dx$>')
plt.xlabel('# throws')

plt.grid(True)
plt.show()
```

# Esercizio 2: Random Walk


```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('data/RW_lattice.csv')

fig=plt.figure(
    figsize=(10,10)
)
ax=fig.add_subplot(projection='3d')
ax.set_box_aspect(None, zoom=0.85)
for run in df['#rw'].unique()[:200]:
    dfr=df.loc[df['#rw']==run]
    ax.plot(dfr.x, dfr.y, dfr.z, '-', label=f'run={run}')
#plt.legend()
ax.set_xlabel('x', fontweight='bold')
ax.set_ylabel('y', fontweight='bold')
ax.set_zlabel('z', fontweight='bold')
#ax.ticks_params(fontweight='bold')
ax.set_title('First 200 Random Walk on a lattice', fontweight='bold', y=0.9)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
dfrn=pd.read_csv('data/RW_rn.csv')
for N in dfrn.N.unique():
    plt.title(f'rN mean value at step {N}')
    dfn=dfrn.loc[dfrn.N==N]
    plt.errorbar(100*dfn.block, dfn.rN, yerr=dfn.err)
    plt.xlabel('# of walks')
    plt.ylabel('<r(N)>')
    plt.show()
```

```python
df=dfrn.loc[dfrn.block==99].reset_index(drop=True)

plt.errorbar(df.N, df.rN, yerr=df.err)
plt.grid(True)
plt.xlabel('N [step]')
plt.ylabel('<r(N)>')
plt.title('Distance explored by 10^4 RWs')
plt.show()
```
