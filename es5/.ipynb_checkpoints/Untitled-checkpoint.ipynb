```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

```

```python
df1s=pd.read_csv('data/pos_1s_gauss.csv')
```

```python
df1s
```

```python

fig = plt.figure()
ax = fig.add_subplot(projection='3d')
ax.scatter(df1s.x, df1s.y, df1s.z, c=df1s.z, marker='.')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
ax.view_init(10, 30)
ax.set_title('Orbital 1s')

plt.show()
```

```python
r1s=pd.read_csv('data/ray_1s_gauss.csv')
r1s
```

```python
fig, ax = plt.subplots(1, 2, figsize=(9,3))

ax[0].set_title('mean ray in orbital 1s')
ax[0].errorbar(r1s.throws, r1s.avg, yerr=r1s.err)
ax[0].grid(True)

ax[1].set_title('error on mean ray for orbital 1s')
ax[1].plot(r1s.throws, r1s.err, '-*')
ax[1].grid(True)
plt.show()
```

```python

fig = plt.figure(figsize=(10,10))
ax = fig.add_subplot(projection='3d')
ax.scatter(df2p.x, df2p.y, df2p.z, c=df2p.z, alpha=0.01, marker='.')
ax.scatter(df1s.x, df1s.y, df1s.z, c='b', alpha=0.1, marker='.')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
ax.view_init(0, 60)

plt.show()
```

```python
df2s=pd.read_csv('data/pos_2s_gauss.csv')
df3s=pd.read_csv('data/pos_3s_gauss.csv')

df2s
```

```python

fig = plt.figure()
ax = fig.add_subplot(projection='3d')
#ax.scatter(df1s.x, df1s.y, df1s.z, c='b', alpha=0.1, marker='.')
ax.scatter(df2s.x, df2s.y, df2s.z, c='r', marker='.', alpha=0.01)
ax.scatter(df3s.x, df3s.y, df3s.z, c=df3s.z, marker='.', alpha=0.001)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
ax.view_init(0, 30)
ax.set_title('Orbital 1s, 2s and 3s')

plt.show()
```
