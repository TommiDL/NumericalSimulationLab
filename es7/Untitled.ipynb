```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python

df=pd.read_csv('NSL_SIMULATOR/OUTPUT/potential_energy.dat',  delimiter=r"\s+")
```

```python
df
```

```python
plt.title('Potential energy')
plt.errorbar(x=df['#BLOCK:'], y=df['PE_AVE:'], yerr=df['ERROR:'], label='Average')
plt.plot(df['ACTUAL_PE:'])

plt.grid(True)
plt.legend()
plt.show()
```

```python
dfp=pd.read_csv('NSL_SIMULATOR/OUTPUT/pressure.dat', delimiter=r"\s+")
dfp
```

```python
plt.title('Pressure')
plt.errorbar(x=dfp['#BLOCK:'], y=dfp['P_AVE:'], yerr=dfp['ERROR:'], label='Average')
plt.plot(dfp['ACTUAL_P:'])

plt.grid(True)
plt.legend()
plt.show()
```
