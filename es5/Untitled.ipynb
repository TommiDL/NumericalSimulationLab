```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

```

# Orbitale 1s

### Gaussian distribution

```python
g1s=pd.read_csv('data/pos_1s_gauss.csv')

fig = plt.figure()
ax = fig.add_subplot(projection='3d')
ax.scatter(g1s.x, g1s.y, g1s.z, c=g1s.z, marker='.')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
ax.view_init(10, 30)
ax.set_title('Orbital 1s gauss distribution')

plt.show()
```

```python
r1s=pd.read_csv('data/ray_1s_gauss.csv')

fig, ax = plt.subplots(1, 2, figsize=(9,3))

ax[0].set_title('mean ray 1s gauss distribution')
ax[0].errorbar(r1s.throws, r1s.avg, yerr=r1s.err)
ax[0].hlines(1.5, 0, ru1s.throws[len(ru1s)-1], color='r', ls='--')
ax[0].grid(True)

ax[1].set_title('error on mean ray 1s gauss distribution')
ax[1].plot(r1s.throws, r1s.err, '-*')
ax[1].grid(True)
plt.show()
```

### Uniform distribution

```python
u1s=pd.read_csv('data/pos_1s_unif.csv')

fig = plt.figure()
ax = fig.add_subplot(projection='3d')
ax.scatter(u1s.x, u1s.y, u1s.z, c=u1s.z, marker='.')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
ax.view_init(10, 30)
ax.set_title('Orbital 1s with uniform distribution')

plt.show()
```

```python
ru1s=pd.read_csv('data/ray_1s_unif.csv')

fig, ax = plt.subplots(1, 2, figsize=(9,3))

ax[0].set_title('mean ray 1s uniform distribution')
ax[0].errorbar(ru1s.throws, ru1s.avg, yerr=ru1s.err)
ax[0].hlines(1.5, 0, ru1s.throws[len(ru1s)-1], color='r', ls='--')
ax[0].grid(True)

ax[1].set_title('error on mean ray 1s uniform distribution')
ax[1].plot(ru1s.throws, ru1s.err, '-*')
ax[1].grid(True)
plt.show()
```
